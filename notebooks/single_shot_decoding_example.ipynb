# Single-shot decoding example

This notebook explains one from-scratch surface-code decoding shot. It shows how an error chain creates syndrome defects and how the decoder returns a correction chain.


```python
from src.lattice import PlanarSurfaceCodeLattice, Edge, xor_edges
from src.decoder import BruteForceMWPMDecoder
from src.visualize import draw_decoding_state
```

```python
lattice = PlanarSurfaceCodeLattice(distance=5, error_type='z')
error_edges = {Edge('h', 2, 1), Edge('v', 1, 3)}
syndrome = lattice.toggle_syndrome_from_edges(error_edges)
decoder = BruteForceMWPMDecoder(lattice)
decoded = decoder.decode(syndrome)
residual = xor_edges(error_edges, decoded.correction_edges)
syndrome, decoded.matches, lattice.logical_parity(residual)
```

```python
draw_decoding_state(lattice, syndrome, decoded.correction_edges, residual, title='Single-shot decoding example')
```
